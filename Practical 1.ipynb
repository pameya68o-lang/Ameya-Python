Electrical formulas used:
    
• Series resistance: Rs = R1 + R2 + ...

• Parallel resistance: 1/Rp = 1/R1 + 1/R2 + ...

• Ohm’s Law: V = I × R

Syntax

• print() for output
          
• input() for user input
          
• float() for numeric conversion

• Arithmetic operators: + − * /

Program to calculate equivalent resistance in series

```python
r1= float(input("r1="))
r2=float(input("r2="))
rs=r1+r2
print(rs)
```

Program to calculate equivalent resistance in parallel

```python
r1=float(input("r1="))
r2=float(input("r2="))
rp=1/(1/r1)+(1/r2)
print(rp)
```
